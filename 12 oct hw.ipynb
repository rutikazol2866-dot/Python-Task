### Find topper in physics subject,maths subject,chemistry subject.

```python
db = {1:[78,97,86],2:[67,89,90],3:[78,90,87]}
subject = ["physics", "maths", "chemistry"]

for i in range(3):
    topper_roll = list(db.keys())[0]
    topper_marks = db[topper_roll][i]
     
    for k,v in db.items():
        if v[i] > topper_marks:
            topper_roll = k
            topper_marks = v[i]


    print(f"{subject[i]} topper roll no is {topper_roll} with marks {topper_marks}")        
```
